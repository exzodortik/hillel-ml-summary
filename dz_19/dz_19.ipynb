```python
import torch
import transformers
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
# Загружаем модель
model_name = 'Helsinki-NLP/opus-mt-en-uk'
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)
```

```python
# Подготавливаем текст
text = 'I enjoy learning new languages and reading books.'
inputs = tokenizer(text, return_tensors='pt')
print(f'{inputs['input_ids']}, {inputs['attention_mask']}')
```

```python
# Генерируем перевод
with torch.inference_mode():
    output_ids = model.generate(**inputs, max_new_tokens=64)
    translation = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    print(translation)
```

```python
# Делаем функцию translate_text, которая выполняется токенизацию, генерацию и декодирование
def translate_text(text):
    inputs = tokenizer(text, return_tensors='pt')
    with torch.inference_mode():
        output_ids = model.generate(**inputs, max_new_tokens=64)
        translation = tokenizer.decode(output_ids[0], skip_special_tokens=True)
    return translation

text2 = 'I visited my friend yesterday.'
print(translate_text(text2))
```

```python
# Еще один пример для проверки, выше почему то модель показала подругу но там нету конкретизации пола
text3 = "If it rains tomorrow, we will stay at home."
print(translate_text(text3))
```

Модель Helsinki-NLP/opus-mt-en-uk використано для перекладу з англійської на українську. У перевірених прикладах зміст збережено, проте модель може конкретизувати стать, не зазначену в оригіналі.
