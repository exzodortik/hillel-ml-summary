```python
import numpy as np
```

Download dataset without target (Iris-setosa, Iris-versicolor, Iris-virginica)

```python
data = np.genfromtxt(
    'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data',
    usecols=(0, 1, 2, 3),
    delimiter=',',
    dtype=np.float64)

total_elements = data.size
```

Check if the data is a 1D or 2D array. Output will be: 1 for a 1D array (vector), 2 for a 2D array (matrix)

```python
print(f'Array type: {data.ndim}')
```

Calculating mean, median, standard deviation for first column

```python
print(f'Mean of first column: {np.mean(data[:, 0])}')
print(f'Median of first column: {np.median(data[:, 0])}')
print(f'Standard deviation of first column: {np.std(data[:, 0])}')
```

Generate 20 unique flat indices without duplicates

```python
unique_flat_indices = np.random.choice(total_elements, size=20, replace=False)
```

Insert nan in unique flat indices using .ravel()

```python
data.ravel()[unique_flat_indices] = np.nan
```

Find inserted nan in first column

```python
print(np.where(np.isnan(data[:, 0])))
```

Replace np.nan with 0

```python
data[np.isnan(data)] = 0
```

Filter the array behind the sink: values in the 3rd column > 1.5 and values in the 1st column < 5.0

```python
filtered_data = data[(data[:, 2] > 1.5) & (data[:, 0] < 5.0)]
print(filtered_data)
```

Divided vertically without absolute numbers

```python
part1, part2 = np.vsplit(data, 2)
```

Sorted first array according to the 1st column by magnification

```python
idx_asc = np.argsort(part1[:, 0])
part1_sorted = part1[idx_asc]
```

Sorted second array according to the 1st column by reducing

```python
idx_desc = np.argsort(part2[:, 0])[::-1]
part2_sorted = part2[idx_desc]
```

Merging two arrays into one

```python
data = np.concatenate((part1_sorted, part2_sorted), axis=1)
print(data)
```

Count all unique values in an array and output them along with their counts.

```python
values, counts = np.unique(data, return_counts=True)
print(f'Unique values: {values},\nLength: {len(counts)}')
```

Find the most frequently occurring value in an array.

```python
max_index = np.argmax(counts)
most_frequent_value = values[max_index]
frequency = counts[max_index]
print(f"The most frequently repeated meaning: {most_frequent_value} (occurs {frequency} count)")
```

Function, that get one array, multiplied all values in the column that were less than the column's average by
    2, and divided the other values by 4.

```python
def transform_by_mean(array: np.ndarray) -> np.ndarray:
    average_value = array.mean(axis=0)
    first_mask = array < average_value
    return np.where(first_mask, array * 2, array / 4)

print(transform_by_mean(data[:, 2]))
```
