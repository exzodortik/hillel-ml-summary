```python
import sys
import subprocess
import spacy
from spacy import displacy
```

```python
subprocess.run(
    [sys.executable, '-m', 'spacy', 'download', 'uk_core_news_sm'],
    check=True
)
```

```python
nlp = spacy.load('uk_core_news_sm')
```

```python
# Обработаем небольшой текст
text = (
    "Олена Коваль приїхала зі Львова до Києва. "
    "У Києві вона відвідала офіс компанії Microsoft."
)

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ent.start_char, ent.end_char, ent.label_)
```

```python
# Выводим сущности прямо в ноутбуке
displacy.render(doc, style='ent', jupyter=True)
```

```python
# Обработаем чуть сложнее текст
text2 = (
    "Тарас Шевченко народився в Моринцях. "
    "Олена працює в компанії Нова пошта у Львові. "
    "У суботу вона поїде до Києва."
)
doc2 = nlp(text2)

displacy.render(doc2, style='ent', jupyter=True)
```

Модель spaCy uk_core_news_sm правильно розпізнала людей, місця й організації в перевірених прикладах. Результати показано через displaCy.
